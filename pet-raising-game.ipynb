# ペット育成ゲーム

## 概要

ペットにご飯をあげたり、遊んだり、寝かせたりしながら育てるゲームです。行動によって満腹度、元気、経験値が変化し、経験値が100になるとレベルアップします。

## 操作方法

コードセルを実行し、ペットの名前を入力します。その後、メニューから数字を選んで行動します。

- `1`: ご飯をあげる
- `2`: 遊ぶ
- `3`: 寝かせる
- `4`: ゲームを終了する

行動するたびに満腹度と元気が少しずつ減ります。どちらかが0になるとゲームオーバーです。

## 制作者のコメント

授業で習った関数、if文、クラスを使って作ったゲームです。クラスを使って、名前やレベル、経験値などをまとめて管理しています。経験値が100になったときのレベルアップ処理を何度も実行して確認しました。今後はペットの画像や進化機能も追加してみたいです。

## 作成時間

約6時間（調査や学習時間を含む）

```python
import time


class Pet:
    def __init__(self, name):
        self.name = name
        self.level = 1
        self.exp = 0
        self.food = 50
        self.energy = 50

    def status(self):
        print("\n========================")
        print("🐾 ペット情報")
        print("========================")
        print(f"名前　　 ：{self.name}")
        print(f"レベル　 ：{self.level}")
        print(f"経験値　 ：{self.exp}/100")
        print(f"満腹度　 ：{self.food}/100")
        print(f"元気　　 ：{self.energy}/100")
        print("========================")

    def eat(self):
        self.food += 20
        if self.food > 100:
            self.food = 100

        self.exp += 10
        if self.exp >= 100:
            self.exp -= 100
            self.level += 1
            print(f"\n🎉 {self.name}はレベルアップした！")

        print(f"\n🍖 {self.name}はご飯を食べた！")

    def play(self):
        if self.energy >= 10:
            self.energy -= 10
            self.exp += 20
            if self.exp >= 100:
                self.exp -= 100
                self.level += 1
                print(f"\n🎉 {self.name}はレベルアップした！")

            print(f"\n🎮 {self.name}と遊んだ！")
        else:
            print(f"\n😢 {self.name}は疲れていて遊べない...")

    def sleep(self):
        self.energy += 30
        if self.energy > 100:
            self.energy = 100

        print(f"\n💤 {self.name}はぐっすり眠った！")

    def update(self):
        self.food = max(0, self.food - 5)
        self.energy = max(0, self.energy - 5)


print("=======================")
print("🐾 ペット育成ゲーム 🐾")
print("=======================")

name = input("ペットの名前を決めてください：")
pet = Pet(name)
start_time = time.time()

while True:
    pet.status()
    print("""
1. ご飯をあげる 🍖
2. 遊ぶ 🎮
3. 寝かせる 💤
4. 終了
""")

    choice = input("選択 → ")
    if choice == "1":
        pet.eat()
    elif choice == "2":
        pet.play()
    elif choice == "3":
        pet.sleep()
    elif choice == "4":
        print("\nまた遊んでね！👋")
        break
    else:
        print("\n⚠️ 1～4を選んでください。")
        continue

    pet.update()
    if pet.food == 0:
        print("\n=======================")
        print(f"💀 {pet.name}はお腹が空きすぎて元気がなくなってしまった...")
        print("🎮 GAME OVER")
        print(f"到達レベル：{pet.level}")
        print(f"プレイ時間：{time.time() - start_time:.1f}秒")
        print("=======================")
        break

    if pet.energy == 0:
        print("\n=======================")
        print(f"💀 {pet.name}は疲れすぎて元気がなくなってしまった...")
        print("🎮 GAME OVER")
        print(f"到達レベル：{pet.level}")
        print(f"プレイ時間：{time.time() - start_time:.1f}秒")
        print("=======================")
        break

    time.sleep(1)
```
